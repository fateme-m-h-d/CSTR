```python
import numpy as np

# create a numpy array
a = np.array([1, 2, 3, 4])
print("a=", a)
```

```python
#matrix
b = np.array([[1, 2], [3, 4]])
print("b=", b)
```

```python
#shape
print(a.shape)
print(b.shape)
```

```python
#reshape
c = a.reshape((4, 1))
print("c=", c)
print(c.shape)

d = a.reshape((2,-1,1))
print(d.shape)
```

```python
# concatenate
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

c = np.concatenate((a, b))
print("c=",c)
print(c.shape)
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

c = a * b
print("c", c)
```

```python
#broadcasting
c = np.array([10])

d = a + c
print("d=", d)


e = a + 10
print ("e", e)
```

```python
#slicing
a = np.array([1, 2, 3, 4, 5, 6, 7, 8])
print(a.shape)

b = a[2:7:2]
print("b=", b)
```

```python
c = a[a>3]
print("c=", c)
```

```python
a = np.array([1, 2, 3, 4, 5, 6, 7, 8])

sum_a = np.sum(a)
print("sum=", sum_a)

max_a = np.max(a)
print("max", max_a)

min_a = np.min(a)
print("min", min_a)
```

```python
#sort
a = np.array([4, 2, 9, 6, 7, 3, 5])
b = np.sort(a)
b
```

```python
#random
a = np.random.randint(-100, 100, (3,4))
a
```

```python
#stacking
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

c = np.vstack((a, b))
print(c.shape)
print("c=", c)
```

```python
d = np.hstack((a, b))
print(d.shape)
print("d=", d)
```

```python
#normalize

a = np.array([1, 2, 3])
mean_a = np.mean(a)
mean_a
```

```python
std_a = np.std(a)
std_a
```

```python
norm_a = (a - mean_a) / std_a
norm_a
```

```python
np.mean(norm_a)
```

```python
np.std(norm_a)
```
